```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('sketch_hitler.png')
if image is None :
    print("Error opening image")
else :
    plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
    plt.title('Hitler')
    plt.axis('off')
```

Min-Max Normalization

```python
min_val = image.min()
max_val = image.max()

min_max_norm = (image - min_val) * (255 / (max_val - min_val))
min_max_norm = min_max_norm.astype(np.uint8)
plt.imshow(cv2.cvtColor(min_max_norm,cv2.COLOR_BGR2RGB))
plt.title('Min-Max')
plt.axis('off')
```

General Normalization

```python
gen_norm = image.astype(np.float32)/255.0     #matplotlib automatically recognizes float images in range 0-1
plt.imshow(cv2.cvtColor(gen_norm,cv2.COLOR_BGR2RGB)) #matplotlib can hence display it correctly while cv2 cannot... it expects uint8 images in range 0-255
plt.title('General')
plt.axis('off')
```

Z-Score standardization

```python
img_float = image.astype(np.float32)
mean_colors = np.mean(img_float, axis = (0,1), keepdims = True)
std_colors = np.std(img_float, axis = (0,1), keepdims = True)

print("Mean colors (BGR): ", mean_colors)
print("Std colors (BGR): ", std_colors)

standardized = (img_float - mean_colors) / std_colors
# To visualize, we need to bring the values back to a displayable range
std_min = standardized.min()
std_max = standardized.max()

print('Standardized image ', standardized )
print('Min value ', std_min )
print('Max value ', std_max )
standardized_norm = (standardized - std_min) * (255 / (std_max - std_min))
standardized_norm = standardized_norm.astype(np.uint8)
plt.imshow(cv2.cvtColor(standardized_norm,cv2.COLOR_BGR2RGB))
plt.title('Standardized_min_max_norm')
plt.axis('off')
```

Data Augmentation

```python
flipped_h = image[:, ::-1] #horizontal flip same as cv2.flip(image,1)
plt.imshow(cv2.cvtColor(flipped_h,cv2.COLOR_BGR2RGB)) #similar things can be done for other flips
plt.title('Flipped Horizontal')
plt.axis('off')
```
